### 30. You have the following marks for five students:
marks = np.array([65, 78, 92, 45, 88])

Write a program that:
1. Displays all students' marks.
2. Calculates the average marks.
3. Finds the highest marks.
4. Finds the lowest marks.
5. Counts students who scored at least 70.
6. Displays the marks of students who passed, assuming the passing score is 50.
7. Sorts the marks in ascending order.

```python
marks = np.array([65, 78, 92, 45, 88])

# 1. Display all marks
print("All marks:", marks)

# 2. Calculate average
print("Average marks:", np.mean(marks))

# 3. Find highest marks
print("Highest marks:", np.max(marks))

# 4. Find lowest marks
print("Lowest marks:", np.min(marks))

# 5. Count students scoring at least 70
count = np.sum(marks >= 70)
print("Students scoring at least 70:", count)

# 6. Display marks of students who passed
passed = marks[marks >= 50]
print("Passing marks:", passed)

# 7. Sort marks in ascending order
sorted_marks = np.sort(marks)
print("Sorted marks:", sorted_marks)
```
